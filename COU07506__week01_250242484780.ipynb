```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from scipy.io import wavfile

plt.rcParams["figure.figsize"] = (9, 3.5)

# Task 1.1: Elementary Discrete-Time Signals

n = np.arange(-5, 11)

# Unit Impulse
delta = (n == 0).astype(float)

# Unit Step
step = (n >= 0).astype(float)

# Unit Ramp
ramp = n * step

# Exponential Signal
expo = 0.8**n * step

# Plot the four signals
fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))

for ax, x, title in zip(
    axes,
    [delta, step, ramp, expo],
    ["Unit Impulse", "Unit Step", "Unit Ramp", "Exponential"]
):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title)
    ax.set_xlabel("n")
    ax.set_ylabel("Amplitude")
    ax.grid(True)

plt.tight_layout()
plt.show()
```

```python
n = np.arange(0, 21)

for a in [0.5, -0.8, 1.1]:
    x = a**n
    plt.stem(n, x)
    plt.title(f"Exponential Signal: a = {a}")
    plt.xlabel("n")
    plt.ylabel("Amplitude")
    plt.grid(True)
    plt.show()
```

```python
n = np.arange(0, 41)

x1 = np.cos(2*np.pi*n/8)
x2 = np.cos(2*np.pi*n/7.5)

fig, axes = plt.subplots(2, 1, figsize=(10, 5))

axes[0].stem(n, x1)
axes[0].set_title("cos(2πn/8)")
axes[0].set_xlabel("n")

axes[1].stem(n, x2)
axes[1].set_title("cos(2πn/7.5)")
axes[1].set_xlabel("n")

plt.tight_layout()
plt.show()
```

```python
x = x.astype(float)
x = x / np.max(np.abs(x))
```
